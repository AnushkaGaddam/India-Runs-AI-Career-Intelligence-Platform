```python
import pandas as pd

df = pd.read_json(r"C:\Users\Gaddam Anushka\Downloads\[PUB] India_runs_data_and_ai_challenge\[PUB] India_runs_data_and_ai_challenge\India_runs_data_and_ai_challenge\candidates.jsonl", lines=True)
```

```python
# Basic profile fields

df["headline"] = df["profile"].apply(lambda x: x.get("headline"))
df["summary"] = df["profile"].apply(lambda x: x.get("summary"))
df["experience"] = df["profile"].apply(lambda x: x.get("years_of_experience"))
df["current_title"] = df["profile"].apply(lambda x: x.get("current_title"))
df["current_company"] = df["profile"].apply(lambda x: x.get("current_company"))
df["country"] = df["profile"].apply(lambda x: x.get("country"))
```

```python
df["skill_names"] = df["skills"].apply(
    lambda skills: ", ".join([s["name"] for s in skills])
)
```

```python
df[["skill_names"]].head()
```

```python
df["skill_details"] = df["skills"].apply(
    lambda skills: ", ".join(
        [f"{s['name']} ({s['proficiency']})" for s in skills]
    )
)
```

```python
df["num_skills"] = df["skills"].apply(len)
```

```python
df[["num_skills"]].head()
```

```python
df["num_jobs"] = df["career_history"].apply(len)
```

```python
df["highest_degree"] = df["education"].apply(
    lambda edu: edu[0]["degree"] if len(edu) > 0 else ""
)
```

```python
df["num_certifications"] = df["certifications"].apply(len)
```

```python
df["language_names"] = df["languages"].apply(
    lambda langs: ", ".join([l["language"] for l in langs])
)
```

```python
df["profile_score"] = df["redrob_signals"].apply(
    lambda x: x["profile_completeness_score"]
)

df["github_score"] = df["redrob_signals"].apply(
    lambda x: x["github_activity_score"]
)

df["response_rate"] = df["redrob_signals"].apply(
    lambda x: x["recruiter_response_rate"]
)

df["connections"] = df["redrob_signals"].apply(
    lambda x: x["connection_count"]
)

df["endorsements"] = df["redrob_signals"].apply(
    lambda x: x["endorsements_received"]
)

df["open_to_work"] = df["redrob_signals"].apply(
    lambda x: x["open_to_work_flag"]
)
```
