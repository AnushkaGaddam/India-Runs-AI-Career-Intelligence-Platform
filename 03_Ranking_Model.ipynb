```python
import pandas as pd

df = pd.read_json(r"C:\Users\Gaddam Anushka\Downloads\[PUB] India_runs_data_and_ai_challenge\[PUB] India_runs_data_and_ai_challenge\India_runs_data_and_ai_challenge\candidates.jsonl", lines=True)
```

```python
AI_SKILLS = {
    "Python",
    "Machine Learning",
    "Deep Learning",
    "NLP",
    "LLM",
    "Fine-tuning LLMs",
    "Embeddings",
    "Vector Databases",
    "Pinecone",
    "Weaviate",
    "Qdrant",
    "Milvus",
    "FAISS",
    "Elasticsearch",
    "OpenSearch",
    "Spark",
    "PySpark",
    "Airflow",
    "SQL",
    "Ranking",
    "Recommendation Systems",
    "Retrieval",
    "Transformers"
}
```

```python
def count_ai_skills(skills):

    names = [s["name"] for s in skills]

    return len(set(names).intersection(AI_SKILLS))

df["ai_skill_score"] = df["skills"].apply(count_ai_skills)
```

```python
def experience_score(x):

    if 5 <= x <= 9:
        return 10

    elif 4 <= x < 5:
        return 8

    elif 9 < x <= 11:
        return 8

    elif 2 <= x < 4:
        return 5

    else:
        return 2

df["experience_score"] = df["profile"].apply(
    lambda x: experience_score(x["years_of_experience"])
)
```

```python
df["open_to_work_bonus"] = df["redrob_signals"].apply(
    lambda x: 5 if x["open_to_work_flag"] else 0
)
```

```python
df["github_score"] = df["redrob_signals"].apply(
    lambda x: max(x["github_activity_score"], 0)
)
```

```python
df["response_score"] = df["redrob_signals"].apply(
    lambda x: x["recruiter_response_rate"] * 10
)
```

```python
df["profile_score"] = df["redrob_signals"].apply(
    lambda x: x["profile_completeness_score"] / 10
)
```

```python
df["final_score"] = (
      df["ai_skill_score"] * 5
    + df["experience_score"] * 3
    + df["github_score"] * 1.5
    + df["response_score"] * 2
    + df["profile_score"] * 1
    + df["open_to_work_bonus"]
)
```

```python
top = df.sort_values(
    "final_score",
    ascending=False
)

top[[
    "candidate_id",
    "final_score"
]].head(20)
```

```python
# Weighted keywords based on importance in the JD
KEYWORD_WEIGHTS = {
    "ranking": 8,
    "retrieval": 8,
    "recommendation": 8,
    "embedding": 7,
    "embeddings": 7,
    "vector": 6,
    "llm": 7,
    "transformer": 6,
    "fine tuning": 6,
    "rag": 6,
    "machine learning": 5,
    "deep learning": 5,
    "nlp": 5,
    "search": 4,
    "spark": 3,
    "airflow": 3,
    "sql": 2,
    "python": 4,
    "pytorch": 4,
    "tensorflow": 4
}

def weighted_ai_score(history):
    score = 0

    for job in history:
        desc = job.get("description", "").lower()

        for keyword, weight in KEYWORD_WEIGHTS.items():
            if keyword in desc:
                score += weight

    return score

df["career_ai_score"] = df["career_history"].apply(weighted_ai_score)
```

```python
KEYWORD_WEIGHTS = {
    "ranking": 8,
    "retrieval": 8,
    "recommendation": 8,
    "embedding": 7,
    "embeddings": 7,
    "vector": 6,
    "llm": 7,
    "transformer": 6,
    "fine tuning": 6,
    "rag": 6,
    "machine learning": 5,
    "deep learning": 5,
    "nlp": 5,
    "search": 4,
    "spark": 3,
    "airflow": 3,
    "sql": 2,
    "python": 4,
    "pytorch": 4,
    "tensorflow": 4
}

def weighted_ai_score(history):

    full_text = " ".join(
        job.get("description", "").lower()
        for job in history
    )

    score = 0

    for keyword, weight in KEYWORD_WEIGHTS.items():
        if keyword in full_text:
            score += weight

    return score

df["career_ai_score"] = df["career_history"].apply(weighted_ai_score)
df["career_ai_score"].describe()
```

```python
def build_candidate_text(row):

    summary = row["profile"].get("summary", "")

    headline = row["profile"].get("headline", "")

    skills = " ".join([s["name"] for s in row["skills"]])

    career = " ".join([job.get("description", "") for job in row["career_history"]])

    return headline + " " + summary + " " + skills + " " + career

df["candidate_text"] = df.apply(build_candidate_text, axis=1)
```
